```python
# 3
LAST_NAME = "BENAVIDEZ"
STUDENT_ID = "TUPM-26-1835"
SEED_DIGIT = int(STUDENT_ID[-1])
FAVORITE_ARTIST = "DILAW"


# ==============================
# STUDENT-SPECIFIC INPUTS
# ==============================

print("=== STUDENT-SPECIFIC INPUTS ===")
print("LAST NAME:", LAST_NAME)
print("STUDENT ID:", STUDENT_ID)
print("SEED DIGIT:", SEED_DIGIT)
print("FAVORITE ARTIST:", FAVORITE_ARTIST)


# ==============================
# GENERATOR
# ==============================

def generate_telemetry():
    base = SEED_DIGIT + len(LAST_NAME) + len(FAVORITE_ARTIST)

    for i in range(1, 11):
        if i == 6:
            yield "INVALID"
        else:
            yield base * i


# ==============================
# VALIDATION
# ==============================

def validate_telemetry(value):
    if not isinstance(value, (int, float)):
        raise ValueError("Invalid telemetry value")

    if value < 0:
        raise ValueError("Negative telemetry value")

    return value


# ==============================
# DECORATOR
# ==============================

def monitor(function):
    def wrapper(*args, **kwargs):
        print("\nProcessing started...")
        result = function(*args, **kwargs)
        print("Processing completed.")
        return result

    return wrapper


# ==============================
# RECURSION
# ==============================

def recursive_analysis(abnormal, index=0, trace=None):

    if trace is None:
        trace = []

    # Base condition
    if index >= len(abnormal):
        return trace

    trace.append(abnormal[index])

    return recursive_analysis(abnormal, index + 1, trace)


# ==============================
# MAIN PROCESSING FUNCTION
# ==============================

@monitor
def process_telemetry():

    valid_count = 0
    invalid_count = 0
    processed_count = 0

    processed_data = []
    abnormal_data = []

    # Lambda transformation
    transform = lambda x: round(x * 1.10, 2)

    print("\n=== GENERATED TELEMETRY DATA ===")

    # Generator processes data one at a time
    for value in generate_telemetry():

        print("Generated:", value)

        try:
            valid_value = validate_telemetry(value)
            valid_count += 1

            processed_value = transform(valid_value)
            processed_data.append(processed_value)

            processed_count += 1

            # Detect abnormal condition
            if processed_value >= 50:
                abnormal_data.append(processed_value)

        except (ValueError, TypeError) as error:
            invalid_count += 1
            print("Invalid reading:", error)

    # Recursive analysis
    recursive_trace = recursive_analysis(abnormal_data)

    # Returned diagnostic report
    return {
        "processed": processed_count,
        "valid": valid_count,
        "invalid": invalid_count,
        "processed_data": processed_data,
        "abnormal": abnormal_data,
        "recursive_trace": recursive_trace
    }


# ==============================
# DIAGNOSTIC REPORT
# ==============================

report = process_telemetry()


print("\n=== VALID / INVALID RESULTS ===")
print("Valid Readings:", report["valid"])
print("Invalid Readings:", report["invalid"])


print("\n=== PROCESSED RESULTS ===")
print("Processed Readings:", report["processed"])
print("Processed Data:", report["processed_data"])


print("\n=== RECURSIVE ANALYSIS ===")
print("Abnormal Conditions:", report["abnormal"])
print("Recursive Trace:", report["recursive_trace"])


print("\n=== FINAL DIAGNOSTIC SUMMARY ===")

if report["abnormal"]:
    status = "ABNORMAL"
else:
    status = "NORMAL"

print("Processed Readings:", report["processed"])
print("Valid Readings:", report["valid"])
print("Invalid Readings:", report["invalid"])
print("Detected Abnormal Conditions:", len(report["abnormal"]))
print("Overall Equipment Status:", status)


print("\n=== EXECUTION LOG ===")
print("Telemetry generated using student-specific inputs.")
print("Generator processed readings one at a time.")
print("Invalid telemetry values were handled.")
print("Lambda transformation applied.")
print("Decorator monitored the processing function.")
print("Recursive analysis completed.")
print("Final diagnostic report generated.")


print("\n=== FINAL OUTPUT ===")
print("Equipment Status:", status)
```
